```python
import matplotlib.pyplot as plt
import cv2
import torch
from torchvision.models import resnet50, ResNet50_Weights
from PIL import Image
import numpy as np
import torchvision.transforms as transforms
import torchvision.models as models
from sklearn.metrics.pairwise import cosine_similarity
from skimage.transform import resize
from skimage.io import imread
from skimage.color import rgb2gray
from skimage.feature import hog
from sklearn.metrics.pairwise import cosine_similarity
from google.colab.patches import cv2_imshow


from google.colab import drive
drive.mount('/content/drive')
import os
```

```python
def extract_feature(image_path):
    image = imread(image_path)
    if image.shape[-1] == 4:
        image = image[..., :3]
    image_resized = resize(image, (128, 64))
    gray = rgb2gray(image_resized)
    features, _ = hog(gray, pixels_per_cell=(8, 8), cells_per_block=(2, 2), visualize=True)
    return features, image
```

```python
dataset_dir = '/content/drive/MyDrive/Colab Notebooks/image/train'
query_path = '/content/drive/MyDrive/Colab Notebooks/image/train/orange_76.jpg'

query_feat, query_img = extract_feature(query_path)
```

```python
results = []
for filename in os.listdir(dataset_dir):
    path = os.path.join(dataset_dir, filename)
    if path.endswith(('.jpg', '.png', '.jpeg')):
        feat, _ = extract_feature(path)
        sim = cosine_similarity([query_feat], [feat])[0][0]
        results.append((filename, sim))

results.sort(key=lambda x: x[1], reverse=True)

top_n = int(input("enter the number of input images"))
cols = 4
rows = (top_n + 1 + cols - 1) // cols
fig, axs = plt.subplots(rows, cols, figsize=(16, 4 * rows))
axs = axs.flatten()


axs[0].imshow(query_img)
axs[0].set_title("Query Image")
axs[0].axis('off')


top_n = min(top_n, len(results))
for i in range(top_n):
    img_path = os.path.join(dataset_dir, results[i][0])
    result_img = imread(img_path)
    axs[i + 1].imshow(result_img)
    axs[i + 1].set_title(f"{results[i][0]}\nSim: {results[i][1]:.2f}")
    axs[i + 1].axis('off')

for j in range(top_n + 1, len(axs)):
    axs[j].axis('off')

plt.tight_layout()
plt.show()
```
